```python
import pandas as pd
```

```python
df = pd.read_csv("processed_clause_dataset.csv")

df.head()
```

```python
import re

def clean_text(text):

    text = str(text)

    text = text.lower()

    text = re.sub(r'\n+', ' ', text)

    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)

    text = re.sub(r'\s+', ' ', text)

    return text.strip()

df["clean_text"] = df["answer_text"].apply(clean_text)
```

```python
df[["answer_text","clean_text"]].head()
```

```python
df["clean_text"].isnull().sum()
```

```python
df["text_length"] = df["clean_text"].str.len()

df["text_length"].describe()
```

```python
df.to_csv(
    "processed_clause_dataset_clean.csv",
    index=False
)
```
